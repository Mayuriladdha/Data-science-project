#### Chart - 1 Distribution of Movies vs TV Shows becomes:

````python
# Chart - 1 Count Movies and TV Shows
plt.figure(figsize=(8,5))
sns.countplot(data=df1, x='type')

plt.title('Distribution of Movies and TV Shows on Amazon Prime')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.show()
````

##### 1. Why did you pick the specific chart?

A Count Plot (Bar Chart) was chosen because the variable 'type' is a categorical variable containing two categories: Movie and TV Show. Bar charts are one of the most effective visualization techniques for comparing the frequency or count of different categories.

The chart provides a clear and easy-to-understand comparison of the number of Movies and TV Shows available on Amazon Prime. It allows us to quickly identify which content type dominates the platform and helps establish a foundation for further content analysis.

Additionally, the bar chart makes it easier to compare absolute values than other chart types and effectively highlights differences in content distribution across categories.

Key Reason for Selection:

Suitable for categorical data.

Clearly compares frequencies between categories.

Easy to interpret and communicate insights.

Helps understand the overall composition of Amazon Prime's content library.

##### 2. What is/are the insight(s) found from the chart?

The chart reveals the distribution of content types available on Amazon Prime Video. By comparing the number of Movies and TV Shows, we can identify which category contributes the largest share of the platform's content library.

Insights:

Amazon Prime's catalog is dominated by one content type (most likely Movies, based on the dataset).

The platform offers a significantly larger collection of Movies compared to TV Shows.

This indicates that Amazon Prime has historically focused more on movie content while maintaining a smaller but diverse collection of TV Shows.

The content distribution reflects Amazon Prime's strategy of providing a broad selection of films to cater to varied audience preferences.


Business Insight:

A higher proportion of Movies suggests that Amazon Prime prioritizes movie-based entertainment. This information can help stakeholders understand content investment patterns and identify opportunities to expand TV Show offerings if audience demand increases.

Conclusion:

The chart provides a clear overview of the platform's content composition and shows that one content type plays a dominant role in Amazon Prime's library, helping guide further analysis of genres, ratings, and audience engagement.

#### Chart - 2

````python
##Chart - 2 Content Release Trend by Year

plt.figure(figsize=(14,6))

df1['release_year'].value_counts().sort_index().plot(kind='line', marker='o')

plt.title('Content Released by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
````

##### 1. Why did you pick the specific chart?

The box plot was chosen to analyze the relationship between user ratings and review length because it effectively shows the distribution and variation of a numerical variable across different categories.

In this case, the chart compares review length (numerical variable) with ratings (categorical variable from 1 to 5) in the dataset from the Google Play Store. A box plot is particularly useful because it displays key statistical measures such as median, quartiles, and outliers, providing a deeper understanding of how review lengths vary for each rating category.

This visualization helps identify whether users with different satisfaction levels (ratings) behave differently in terms of how much feedback they provide. It is especially useful in detecting patterns, such as whether dissatisfied users tend to write longer and more detailed reviews compared to satisfied users.

##### 2. What is/are the insight(s) found from the chart?

From the box plot showing the relationship between review length and ratings in the dataset from the Google Play Store, the following insights were observed:

Lower ratings (1 and 2) tend to have longer reviews, indicating that dissatisfied users usually provide detailed feedback about the issues they face.

Higher ratings (4 and 5) generally have shorter reviews, suggesting that satisfied users often give quick and brief responses.

There is a higher variation in review length for low ratings, which shows that negative experiences lead to more expressive and detailed feedback.

The presence of outliers (very long reviews) across ratings indicates that some users provide extensive feedback regardless of their satisfaction level.

Key Insight

Dissatisfied users are more likely to write detailed and lengthy reviews, while satisfied users tend to keep their feedback short and concise.

Yes, the insights from the review length vs rating analysis of the dataset from the Google Play Store can create a meaningful positive business impact.

Businesses can use this detailed feedback to identify and fix critical issues, leading to improved app quality and higher user satisfaction.

By focusing on these detailed complaints, companies can make targeted improvements, which can increase ratings, enhance user retention, and boost app downloads.

However, there are also insights that may lead to negative growth if ignored:
